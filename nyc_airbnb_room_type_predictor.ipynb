```python
X=df_clean.drop(columns=['room_type'])
y = df_clean['room_type']
```

```python
#stratisfy
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.33, random_state=42,stratify=y)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder,PowerTransformer
from sklearn.compose import ColumnTransformer
```

```python
skewed_cols = ["price", "minimum_nights", "number_of_reviews",
               "reviews_per_month", "calculated_host_listings_count",
               "availability_365"]
geo_cols = ["latitude", "longitude"]
categorical_cols = ['neighbourhood_group', 'neighbourhood']

skew_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy="median")),
    ('skewness', PowerTransformer(method='yeo-johnson')),
    ('scale', StandardScaler())
])
geo_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy="median")),
    ('scale', StandardScaler())
])
categorical_pipeline = Pipeline(steps=[
    ('impute', SimpleImputer(strategy="most_frequent")),
    ('encode', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', skew_pipeline, skewed_cols),
    ('geo', geo_pipeline, geo_cols),
    ('cat', categorical_pipeline, categorical_cols)
])
```

```python
import warnings
warnings.filterwarnings('ignore')


from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble  import RandomForestClassifier,GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
```

```python
# class_weight="Balanced" tells the model to pay more attention to the minority class
# or rare class (Shared Room)
# but the problem is GradintBossting doesnt support class weight so we left it as it is.

models = {
    "Logistic Regression" : LogisticRegression(class_weight="balanced",random_state=42),
    "Decision Tree"       : DecisionTreeClassifier(class_weight="balanced",random_state=42),
    "Random Forest"       : RandomForestClassifier(class_weight="balanced", random_state=42),
    "Gradient Boosting"   : GradientBoostingClassifier(random_state=42),
}

for name,model in models.items():
    pipe=Pipeline(steps=[("preprocessor",preprocessor),("classifier",model)])
#accuracy alone can be misleading on imbaklance classes so we uses macro f1 both planbe accuracy and macro f2 score which treats every class equally
    accuracy = cross_val_score(pipe, X_train, y_train,cv=3,scoring="accuracy")
    macrof1 = cross_val_score(pipe, X_train, y_train,cv=3,scoring="f1_macro")
    print(f"{name} -> Accuracy: {accuracy.mean():.3f} , F1: {macrof1.mean():.3f}")


```

```python
from sklearn.model_selection import RandomizedSearchCV
#hyper parameter tuning
best_pipeline=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("classifier",RandomForestClassifier(class_weight="balanced"))
])

param_distribution = {
    "classifier__n_estimators": [100, 150, 200, 300],
    "classifier__max_depth": [8, 12, 15, 20, None],
    "classifier__min_samples_split": [2, 5, 10],
}

search=RandomizedSearchCV(
    estimator=best_pipeline,
    param_distributions=param_distribution,
    n_iter=10,
    cv=3,
    scoring="f1_macro",
    random_state=42
    )

search.fit(X_train,y_train)
```

```python
print("Best Parameter",search.best_params_)
print("Best vd macro f1",search.best_score_)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix,f1_score
best_pipeline = search.best_estimator_

y_pred = best_pipeline.predict(X_test)

print(f"Accuracy Score: {accuracy_score(y_test, y_pred)}")
print(f"F1 Score: {f1_score(y_test, y_pred, average="macro")}")
```

```python
import numpy as np
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.metrics import f1_score, accuracy_score, recall_score, classification_report

classes = np.unique(y_train)
si = list(classes).index("Shared room")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

def best_shared_weight(proba, y_true):
    """Find how much to boost Shared room's probability, using train data only."""
    best = (1.0, f1_score(y_true, classes[proba.argmax(1)], average="macro"))
    for w in np.arange(1.0, 10.01, 0.25):
        p = proba.copy(); p[:, si] *= w
        f = f1_score(y_true, classes[p.argmax(1)], average="macro")
        if f > best[1]:
            best = (float(w), f)
    return best

def evaluate(name, clf):
    pipe = Pipeline([("preprocessor", preprocessor), ("classifier", clf)])
    oof = cross_val_predict(pipe, X_train, y_train, cv=skf,
                            method="predict_proba", n_jobs=-1)
    w, f_oof = best_shared_weight(oof, y_train)

    pipe.fit(X_train, y_train)
    proba = pipe.predict_proba(X_test)
    default = classes[proba.argmax(1)]
    q = proba.copy(); q[:, si] *= w
    tuned = classes[q.argmax(1)]

    print(f"\n=== {name} (Shared room weight = {w}) ===")
    for label, pred in [("default", default), ("tuned", tuned)]:
        rec = recall_score(y_test, pred, labels=["Shared room"], average=None)[0]
        print(f"{label:8s} acc={accuracy_score(y_test, pred):.3f}  "
              f"macroF1={f1_score(y_test, pred, average='macro'):.3f}  "
              f"sharedRecall={rec:.2f}")
    return pipe, w

rf_pipe, rf_w = evaluate("Random Forest", RandomForestClassifier(
    n_estimators=150, min_samples_split=10,
    class_weight="balanced", random_state=42))

lgbm_pipe, lgbm_w = evaluate("LightGBM", LGBMClassifier(
    n_estimators=400, learning_rate=0.05, num_leaves=63,
    subsample=0.8, subsample_freq=1, colsample_bytree=0.8,
    random_state=42, verbose=-1))
```

```python
proba = lgbm_pipe.predict_proba(X_test)
pred = classes[proba.argmax(1)]
p_shared = proba[:, si]
is_shared = (y_test.values == "Shared room")

for t in [0.10, 0.15, 0.20, 0.25, 0.30, 0.40]:
    flagged = (pred != "Shared room") & (p_shared >= t)
    n = flagged.sum()
    hit = (flagged & is_shared).sum()
    print(f"t={t:.2f}  flagged={n:4d}  actually shared={hit:3d}  precision={hit/max(n,1):.2f}")

print("base rate of Shared room:", round(is_shared.mean(), 3))
```

```python
from features import ListingFeatures, NEW_FEATURES
from lightgbm import LGBMClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report

extra_pipeline = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler()),
])

columns = ColumnTransformer([
    ("num",   skew_pipeline,        skewed_cols),
    ("geo",   geo_pipeline,         geo_cols),
    ("extra", extra_pipeline,       NEW_FEATURES),
    ("cat",   categorical_pipeline, categorical_cols),
])

preprocessor_fe = Pipeline([("features", ListingFeatures()), ("columns", columns)])

fe_pipe = Pipeline([
    ("preprocessor", preprocessor_fe),
    ("classifier", LGBMClassifier(
        n_estimators=400, learning_rate=0.05, num_leaves=63,
        subsample=0.8, subsample_freq=1, colsample_bytree=0.8,
        random_state=42, verbose=-1)),
])

fe_pipe.fit(X_train, y_train)
pred_fe = fe_pipe.predict(X_test)
print("Accuracy:", round(accuracy_score(y_test, pred_fe), 4))
print("Macro F1:", round(f1_score(y_test, pred_fe, average="macro"), 4))
print(classification_report(y_test, pred_fe))
```
